Q1. String Compression
Question

Compress a string by counting consecutive repeated characters.

```python
text = "aaabbccccd"
```

```python
result = ''
count = 1

for x in range(1,len(text)):
    if text[x] == text[x-1]:
        count += 1
    else: 
        result += text[x-1] + str(count)
        count = 1
result += text[-1] + str(count)

print(result)
```

Q2. Count Consecutive Characters



Question

Find the longest consecutive sequence of the same character.

```python
text = "aabbbccccdd"
```

```python
current_count = 1
max_count = 1
max_char = text[0]


for x in range(1,len(text)):
    if text[x] == text[x-1]:
        current_count += 1
    else:
        current_count = 1

    if current_count > max_count:
        max_count = current_count
        max_char = text[x]
print(max_count,max_char)
```

Q4. Longest Word Without max()


Question

Find the longest word without using:

```python
sentence = "Python is extremely powerful"

word = sentence.split()

largest = word[0]

for x in word:
    if len(x) > len(largest):
        largest = x

print(largest)
```
